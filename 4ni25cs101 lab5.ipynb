```python
import numpy as np
import pandas as pd


data = {
    "student":["A","B","C","D","E","F","G","H","I","J"],
    "Marks":[50,55,60,62,65,68,70,72,75,150]
}
df=pd.DataFrame(data)
print(df)
```

```python
Q1=df["Marks"].quantile(0.25)
print(Q1)
```

```python
Q3=df["Marks"].quantile(0.75)
print(Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_limit = Q1 -1.5*IQR
upper_limit = Q3 +1.5*IQR
print(lower_limit)
print(upper_limit)
```

```python
outliers = df[(df['Marks'] < lower_limit) |
              (df['Marks'] > upper_limit)]
```

```python
print("\noutliers using IQR:")
print(outliers)
```

```python
df_capped = df.copy()

df_capped["Marks"] = df_capped["Marks"].clip( 
    lower = lower_limit,
    upper = upper_limit
)
print("\noutlier usinf capping")
print(df_capped)
```
